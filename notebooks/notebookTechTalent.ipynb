```python
# ==========================================
# 1. API & Web Scraping (Data Collection)    data/processed/survey_data_updated 5.zip
# ==========================================
import requests
import pandas as pd
from bs4 import BeautifulSoup

import os

# Create folder if it does not exist yet
os.makedirs("datos_procesados", exist_ok=True)

# Scraping popular programming languages
url_languages = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html"
response = requests.get(url_languages)
soup = BeautifulSoup(response.text, "html.parser")

languages_data = []
for row in soup.find("table").find_all("tr")[1:]:
    cols = row.find_all("td")
    languages_data.append({
        "Language": cols[1].text.strip(),
        "Average_Annual_Salary": cols[3].text.strip()
    })

df_languages = pd.DataFrame(languages_data)
print("Languages Data Collected:")
print(df_languages.head())


# ==========================================
# 2. Data Cleaning & Preprocessing
# ==========================================
file_path = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/VYPrOu0Vs3I0hKLLjiPGrA/survey-data-with-duplicate.csv"
df_survey = pd.read_csv(file_path)

# Remove duplicates
df_survey.drop_duplicates(inplace=True)

# Age Mapping
age_mapping = {
    "Under 18 years old": 17, "18-24 years old": 21, "25-34 years old": 29,
    "35-44 years old": 39, "45-54 years old": 49, "55-64 years old": 59,
    "65 years or older": 70
}
df_survey['Age_numeric'] = df_survey['Age'].map(age_mapping)

# Save cleaned dataset
df_survey.to_csv("datos_procesados/survey_data_updated.csv", index=False)
print(f"Cleaned dataset saved. Shape: {df_survey.shape}")
```
